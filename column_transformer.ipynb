```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing  import OneHotEncoder
from sklearn.impute import SimpleImputer
```

```python
df=pd.read_csv("covid_toy.csv")
df.head()
```

```python
df.info()
# so we have missing values in fever use simple imputer
# for gender,city wwe use one hot encoding
# for cough we will use ordinal encoding
```

```python
from sklearn.compose import ColumnTransformer
transformer=ColumnTransformer(transformers=[
    ('tn1' ,SimpleImputer(),['fever']),
    ('tn2' ,OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
    ('tn3',OneHotEncoder(drop='first'),['gender','city'])
],remainder='passthrough')
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(df.drop('has_covid',axis=1),df['has_covid'],test_size=0.2,random_state=42)
```

```python
df1=pd.DataFrame(transformer.fit_transform(x_train))
df1.shape
# here we can see the 2 more colums are add it is due to one hot encoding
```
