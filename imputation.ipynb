```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("Titanic-Dataset.csv")
df.head()
```

```python
df1=df[['Age','Fare']]
df1.head()
```

```python
from sklearn.impute import SimpleImputer
df1['Age_99']=df1['Age']
df1['Fare_99']=df1['Fare']
df1['Age_minus']=df1['Age']
df1['Fare_minus']=df1['Fare']
```

```python
df1.head()
```

```python
imputer1=SimpleImputer(strategy='constant',fill_value=99)
imputer2=SimpleImputer(strategy='constant',fill_value=-1)
columns=df1.columns 
```

```python
from sklearn.compose import ColumnTransformer
trf=ColumnTransformer(transformers=[
    ('99',imputer1,['Age_99','Fare_99']),
    ('-1',imputer2,['Age_minus','Fare_minus'])
],remainder='passthrough')
```

```python
new_columns = ['Age_99', 'Fare_99', 'Age_minus', 'Fare_minus', 'Age', 'Fare']
df1=trf.fit_transform(df1)
```

```python
df1=pd.DataFrame(df1,columns=new_columns)
```
