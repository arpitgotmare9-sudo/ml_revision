```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("customer (1).csv")
df.head()
```

```python
df = df.iloc[:,2:]
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(df.drop('purchased',axis=1),df['purchased'],test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import OrdinalEncoder
ord=OrdinalEncoder(categories=[['Poor','Average','Good'],['School','UG','PG']])
ord.fit(x_train)
x_test_ord=ord.transform(x_test)
x_train_ord=ord.transform(x_train)
```

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
lr_ord=LogisticRegression()
lr_ord.fit(x_train_ord,y_train)
y_pread_ord=lr_ord.predict(x_test_ord)
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test,y_pread_ord))
```
