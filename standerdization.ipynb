```python
import pandas as pd
```

```python
df=pd.read_csv("Social_Network_Ads (1).csv")
df.head()
df=df.iloc[:,2:]
```

```python
# Their is difference between the age and salery 
df[['Age','EstimatedSalary']].describe()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(df.drop('Purchased',axis=1),df['Purchased'],random_state=0,test_size=0.2)
```

```python
scaler=StandardScaler()
scaler.fit(x_train)
x_test_scaled=scaler.transform(x_test)
x_train_scaled=scaler.transform(x_train)
```

```python
print(x_train.describe())
print(pd.DataFrame(x_train_scaled,columns=x_train.columns).describe())
```

```python
from sklearn.linear_model import LogisticRegression
m=LogisticRegression()
m_scaled=LogisticRegression ()
m.fit(x_train,y_train)
m_scaled.fit(x_train_scaled,y_train)
y_pread=m_scaled.predict(x_test)
y_pread_scaled=m_scaled.predict(x_test_scaled)
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test,y_pread))
print(accuracy_score(y_test,y_pread_scaled))
```
